**CONFIGURANDO UM DIAGNÓSTICO PARA OS DATAFREME**



```python
def diagnose_dataframe(df, df_name):
    print(f"\n--- Diagnóstico para o DataFrame: {df_name} ---")
    print("\nDimensões do DataFrame (Linhas, Colunas):")
    print(df.shape)

    print("\nInformações Básicas (dtype, non-null count):")
    df.info()

    print("\nValores Nulos por Coluna:")
    print(df.isnull().sum())

    print("\nEstatísticas Descritivas para Colunas Numéricas:")
    print(df.describe())

    print("\nNúmero de Valores Únicos por Coluna:")
    print(df.nunique())

# Aplica a função de diagnóstico a cada DataFrame
diagnose_dataframe(df_analistas, 'df_analistas')
diagnose_dataframe(df_projetos, 'df_projetos')
diagnose_dataframe(df_clientes, 'df_clientes')
diagnose_dataframe(df_satisfacao, 'df_satisfacao')
diagnose_dataframe(df_apontamentos, 'df_apontamentos')
```
