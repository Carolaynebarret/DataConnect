```python
#convertendo os meses
meses = {
    "jan": "01",
    "fev": "02",
    "mar": "03",
    "abr": "04",
    "mai": "05",
    "jun": "06",
    "jul": "07",
    "ago": "08",
    "set": "09",
    "out": "10",
    "nov": "11",
    "dez": "12"
}
```

```python
#Funsão para conversão do formato da data
def padronizar_data(data):

    if pd.isna(data):
        return pd.NaT

    data = str(data).strip().lower()

    # Datas com mês escrito: 06-jan-2023
    for mes, numero in meses.items():
        data = data.replace(f"-{mes}-", f"-{numero}-")

    # AAAA-MM-DD
    if len(data) == 10 and data[4] == "-":
        return pd.to_datetime(data, format="%Y-%m-%d", errors="coerce")

    # Datas com /
    if "/" in data:
        partes = data.split("/")

        if len(partes) == 3:
            primeiro, segundo, ano = partes

            # Se o segundo número > 12, necessariamente é MM/DD/YYYY
            if int(segundo) > 12:
                return pd.to_datetime(data, format="%m/%d/%Y", errors="coerce")

            # Caso contrário, assumimos padrão brasileiro DD/MM/YYYY
            return pd.to_datetime(data, format="%d/%m/%Y", errors="coerce")

    # Datas com hífen depois da conversão dos meses
    return pd.to_datetime(data, format="%d-%m-%Y", errors="coerce")
```

```python
#convertendo os meses
meses = {
    "jan": "01",
    "fev": "02",
    "mar": "03",
    "abr": "04",
    "mai": "05",
    "jun": "06",
    "jul": "07",
    "ago": "08",
    "set": "09",
    "out": "10",
    "nov": "11",
    "dez": "12"
}
```

```python
#Função para conversão do formato da data
def padronizar_data(data):

    if pd.isna(data):
        return pd.NaT

    original_data_str = str(data).strip().lower()

    # Tentativa de parsear formatos gerais, incluindo ISO e com tempo
    try:
        return pd.to_datetime(original_data_str, format='mixed', errors='raise', dayfirst=True)
    except (ValueError, TypeError):
        pass # Se falhar, continua para a lógica específica

    # Datas com mês escrito: 06-jan-2023
    processed_data_str = original_data_str
    for mes, numero in meses.items():
        processed_data_str = processed_data_str.replace(f"-{mes}-", f"-{numero}-")

    # AAAA-MM-DD
    if len(processed_data_str) == 10 and processed_data_str[4] == "-":
        return pd.to_datetime(processed_data_str, format="%Y-%m-%d", errors="coerce")

    # Datas com /
    if "/" in processed_data_str:
        partes = processed_data_str.split("/")

        if len(partes) == 3:
            primeiro, segundo, ano = partes

            # Se o segundo número > 12, necessariamente é MM/DD/YYYY
            if int(segundo) > 12:
                return pd.to_datetime(processed_data_str, format="%m/%d/%Y", errors="coerce")

            # Caso contrário, assumimos padrão brasileiro DD/MM/YYYY
            return pd.to_datetime(processed_data_str, format="%d/%m/%Y", errors="coerce")

    # Datas com hífen depois da conversão dos meses
    return pd.to_datetime(processed_data_str, format="%d-%m-%Y", errors="coerce")
```

Isso nos permite aplicar a função `padronizar_data` (agora corrigida) a todos os valores originais, em vez de valores já afetados por tentativas de conversão anteriores.

Também vamos reaplicar a limpeza e conversão da coluna `valor_contrato`.

```python
# 1. Recarregar df_projetos do arquivo CSV para ter os dados originais
df_projetos = pd.read_csv('dc_projetos.csv')

# 2. Criar a coluna 'data_inicio_original' a partir da 'data_inicio' original (string)
df_projetos["data_inicio_original"] = df_projetos["data_inicio"].astype(str)

# 3. Aplicar a função padronizar_data a todas as colunas de data
df_projetos["data_inicio"] = (
    df_projetos["data_inicio_original"]
    .apply(padronizar_data)
)
df_projetos["data_fim_prevista"] = (
    df_projetos["data_fim_prevista"]
    .apply(padronizar_data)
)
df_projetos["data_fim_real"] = (
    df_projetos["data_fim_real"]
    .apply(padronizar_data)
)

# 4. Limpar e converter a coluna valor_contrato para float (reaplicar)
df_projetos['valor_contrato'] = (
    df_projetos['valor_contrato']
    .astype(str)
    .str.replace('R$', '', regex=False)
    .str.replace('.', '', regex=False)
    .str.replace(',', '.', regex=False)
    .str.strip()
)
df_projetos['valor_contrato'] = pd.to_numeric(df_projetos['valor_contrato'], errors='coerce')

# 5. Converter colunas de texto genéricas para string (reaplicar)
colunas_texto = ['projeto_id', 'cliente_id', 'nome_projeto', 'tipo_servico', 'squad', 'status']
df_projetos[colunas_texto] = df_projetos[colunas_texto].astype('string')


print("df_projetos recarregado e datas e valor_contrato padronizados.")

# Exibir as primeiras 10 linhas para verificar as mudanças
print("\nPrimeiras 10 linhas do df_projetos:")
display(df_projetos.head(10))

# Verificar os tipos de dados novamente
print("\nInformações do DataFrame df_projetos após as padronizações:")
df_projetos.info()
```
