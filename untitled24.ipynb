A coluna possui quatro padrões diferentes:

- Vamos criar um cópia do Dataframe para realizar a conversão sem correr o risco de estragar os dados do arquivo original.

```python
#convertendo em sem altera o datafreme original

df_clientes["data_cadastro_original"] = df_clientes["data_cadastro"]
df_clientes["data_cadastro"] = pd.to_datetime(
    df_clientes["data_cadastro"],
    errors="coerce"
)
```

```python
#convertendo os meses
meses = {
    "jan": "01",
    "fev": "02",
    "mar": "03",
    "abr": "04",
    "mai": "05",
    "jun": "06",
    "jul": "07",
    "ago": "08",
    "set": "09",
    "out": "10",
    "nov": "11",
    "dez": "12"
}
```

```python
#Funsão para conversão do formato da data
def padronizar_data(data):

    if pd.isna(data):
        return pd.NaT

    data = str(data).strip().lower()

    # Datas com mês escrito: 06-jan-2023
    for mes, numero in meses.items():
        data = data.replace(f"-{mes}-", f"-{numero}-")

    # AAAA-MM-DD
    if len(data) == 10 and data[4] == "-":
        return pd.to_datetime(data, format="%Y-%m-%d", errors="coerce")

    # Datas com /
    if "/" in data:
        partes = data.split("/")

        if len(partes) == 3:
            primeiro, segundo, ano = partes

            # Se o segundo número > 12, necessariamente é MM/DD/YYYY
            if int(segundo) > 12:
                return pd.to_datetime(data, format="%m/%d/%Y", errors="coerce")

            # Caso contrário, assumimos padrão brasileiro DD/MM/YYYY
            return pd.to_datetime(data, format="%d/%m/%Y", errors="coerce")

    # Datas com hífen depois da conversão dos meses
    return pd.to_datetime(data, format="%d-%m-%Y", errors="coerce")
```

```python
df_clientes["data_cadastro"] = (
    df_clientes["data_cadastro_original"]
    .apply(padronizar_data)
)
```
